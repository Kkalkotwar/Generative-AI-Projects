```python
import gradio as gr
from transformers import pipeline
```

```python
# Load the translation pipeline using the English-to-Hindi MarianMT model
translator = pipeline("translation_en_to_hi", model="Helsinki-NLP/opus-mt-en-hi")
```

```python
# Define the translation function
def translate_text(text):
    # Translate text to Hindi
    translation = translator(text)
    return translation[0]['translation_text']
```

```python

# Set up Gradio interface
interface = gr.Interface(
    fn=translate_text,
    inputs="text",
    outputs="text",
    title="English to Hindi Translation",
    description="Enter English text, and this tool will translate it to Hindi using a lightweight open-source LLM."
)

# Launch the interface
interface.launch()
```
